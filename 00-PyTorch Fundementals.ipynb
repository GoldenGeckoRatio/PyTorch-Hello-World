### One of the common errors

Shape errors

Two rules for matrix multiplication
1. The 2 inner dimensions need to match:
2. The resulting matrix has the shape of the outer dimensions

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1,2],
                        [3,4],
                        [5,6]])
tensor_B = torch.tensor([[7,10],
                        [8,11],
                        [9,12]])

torch.mm(tensor_A, tensor_B)
```

```python
# To fix the tensor shape issues, we can manipulate the shape using transpose
torch.matmul(tensor_A,tensor_B.T)
```

## Reshaping, stacking, squeezing and unsqueezing tensors
* Reshaping - reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keeps the same memory
* stacking - combine multiple tensors on top of each other
* Squeeze - removes all `1` dimensions from a tensor
* Unsequeeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
# Let's create a tensor
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(1,9)
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(1,9)
z, z.shape
```

```python
# Changing z changes x (because a view of a tensor shares the same memory as the original)
z[:,0] = 5
z, x
```

```python
#stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
```

```python
x_reshaped
```

```python
x_reshaped.squeeze()
```

```python
# torch.squeeze() - removes all single dimensions from a target tensor
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimensions from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# Torch.unsqueeze() - adds a single dimension to a target tensor at a specific dim
print(f"\nPrevious target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

# Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearranges the dimensions of a target tensor
x_original = torch.rand(size=(224, 224, 3)) # [height, width, color)_channels]
```

```python
#Permute the original tensor to rearrange the axis
x_permuted = x_original.permute(2, 0, 1) # Shifts axis 0 -1, 1_2, 2->0
print(x_permuted.shape)
print(x_original.shape)
```

## Reproducbility (trying to take random out of random)

In short how a neural network learns:

`start with random numbers -> tensor operations -. update random number to try and make them better representations of the data -> again -> again -> again -> again...

```python
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reporducible tensors

#set the random seed
RANDOM_SEED = 49
torch.manual_seed(RANDOM_SEED)
random_tensor_A = torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
random_tensor_B = torch.rand(3,4)
print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```
